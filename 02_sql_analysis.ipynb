# 02 — SQL Analysis

Load the actual cleaned dataset. This notebook uses SQLite and prints query outputs.

```python
import sqlite3
from pathlib import Path
import pandas as pd
path = Path('../data/cleaned_launches.csv')
if not path.exists():
    raise FileNotFoundError('Create data/cleaned_launches.csv from the course dataset and ensure it has launch_site, landing_success, payload_mass_kg, launch_year columns.')
df = pd.read_csv(path)
con = sqlite3.connect(':memory:')
df.to_sql('spacex_launches', con, index=False, if_exists='replace')
print(df.shape)
df.head()
```

## Launch count by site

```python
pd.read_sql_query('SELECT launch_site, COUNT(*) AS launch_count FROM spacex_launches GROUP BY launch_site ORDER BY launch_count DESC', con)
```

## Landing outcome counts by site

```python
pd.read_sql_query('SELECT launch_site, landing_success, COUNT(*) AS record_count FROM spacex_launches GROUP BY launch_site, landing_success ORDER BY launch_site, landing_success', con)
```

## Payload summary by outcome

```python
pd.read_sql_query('SELECT landing_success, COUNT(*) AS n, AVG(payload_mass_kg) AS avg_payload_kg, MIN(payload_mass_kg) AS min_payload_kg, MAX(payload_mass_kg) AS max_payload_kg FROM spacex_launches WHERE payload_mass_kg IS NOT NULL GROUP BY landing_success', con)
```
