# 03 — Folium Launch-Site Map

Use verified coordinates from your course material or authoritative source. The code intentionally does not invent coordinates.

```python
import pandas as pd
import folium
from pathlib import Path
sites_path = Path('../data/verified_launch_sites.csv')
if not sites_path.exists():
    raise FileNotFoundError('Create data/verified_launch_sites.csv with columns: name, latitude, longitude, launch_count.')
sites = pd.read_csv(sites_path)
m = folium.Map(location=[29.5, -95.0], zoom_start=4)
for _, row in sites.iterrows():
    folium.Marker([row['latitude'], row['longitude']], tooltip=row['name'],
                  popup=f"{row['name']}<br>Launch count: {row['launch_count']}").add_to(m)
Path('../outputs').mkdir(exist_ok=True)
m.save('../outputs/spacex_launch_sites_map.html')
m
```
